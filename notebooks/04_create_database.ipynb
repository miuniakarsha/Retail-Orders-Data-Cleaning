#create SQLite Database

This notebook loads the cleaned retail orders dataset and stores it inside a SQLite database.

The database will be used for SQL analysis and Power BI visualization.

```python
import pandas as pd
import sqlite3
from pathlib import Path
```

```python
cleaned_data_path = Path("../data/processed/orders_cleaned.csv")

df = pd.read_csv(cleaned_data_path)

print("Cleaned dataset loaded successfully")
print("Shape:", df.shape)

df.head()
```

```python
database_folder = Path("../database")
database_folder.mkdir(parents=True, exist_ok=True)

database_path = database_folder / "orders_cleaned.sqlite"

print("Database path:", database_path)
```

```python
connection = sqlite3.connect(database_path)

df.to_sql(
    name="cleaned_orders",
    con=connection,
    if_exists="replace",
    index=False
)

connection.close()

print("SQLite database created successfully")
```

```python
connection = sqlite3.connect(database_path) #check if table was createdd

tables = pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table';",
    connection
)

connection.close()

tables
```

```python
connection = sqlite3.connect(database_path) #check row count from database

row_count = pd.read_sql_query(
    "SELECT COUNT(*) AS total_rows FROM cleaned_orders;",
    connection
)

connection.close()

row_count
```

```python
connection = sqlite3.connect(database_path) #preview databse table

preview = pd.read_sql_query(
    "SELECT * FROM cleaned_orders LIMIT 5;",
    connection
)

connection.close()

preview
```
